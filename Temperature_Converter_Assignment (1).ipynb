# Temperature Conversion and Threshold Check


```python
# Define a reusable function that converts Celsius to Fahrenheit.
def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32

# Set the Celsius threshold used to classify the entered temperature.
THRESHOLD_C = 25

# Ask for one temperature and convert it using the function.
temperature_c = float(input("Enter a temperature in °C: "))
temperature_f = celsius_to_fahrenheit(temperature_c)

# Display the converted temperature.
print(f"{temperature_c:.1f}°C = {temperature_f:.1f}°F")

# Compare the entered temperature with the threshold.
if temperature_c > THRESHOLD_C:
    print(f"This is above the {THRESHOLD_C}°C threshold.")
elif temperature_c < THRESHOLD_C:
    print(f"This is below the {THRESHOLD_C}°C threshold.")
else:
    print(f"This is exactly at the {THRESHOLD_C}°C threshold.")

# Prepare sample temperatures, then convert each one in a loop.
sample_temperatures_c = [0, 10, 20, 30]
print("\nSeveral temperature conversions:")
for temp_c in sample_temperatures_c:
    temp_f = celsius_to_fahrenheit(temp_c)
    print(f"{temp_c}°C = {temp_f:.1f}°F")
```
